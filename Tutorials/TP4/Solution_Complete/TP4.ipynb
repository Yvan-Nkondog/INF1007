## Question 1.
Pour cette question, vous devez retourner les valeurs maximales de 'Happiness score' pour chacune des 10 régions ('Region'). Ainsi la fonction ci-dessous doit retourner un dictionnaire avec ces scores maximaux en ordre décroissant et leurs pays associés en tant que clé. Par exemple: Australia:7.3, Israel:7.21, etc. N'oubliez pas que vous avez accès aux librairies de pandas et numpy qui possèdent plein d'outils pour automatiser un processus de ce genre!

### 1.1 
Vous devez en premier lieu créer une fonction simple qui retourne une base de données de la région voulue, soit celle donnée en paramètre à la fonction CreateSubsetPerRegion(). Un exemple de ce qui doit être retourné est présenté ci-dessous:


```python
def CreateSubsetPerRegion(df, region):
    #TODO Extraire les sous-données par région à l'aide du dataframe pandas ('subset' de données)
   
    filtered_df = df[df['Region'] == region]
    return filtered_df

display(CreateSubsetPerRegion(df,'East Asia'))
display(CreateSubsetPerRegion(df,'Central and Eastern Europe'))
```

### 1.2
Il faut maintenant retourner le dictionnaire ordonné en ordre décroissant du pays associé à la valeur maximale par région tel qu'indiqué dans la question 1. À nouveau, n'oubliez pas que vous avez accès à des librairies qui automatisent ces processus plus facilement. 

```python
def ClassMaxScorePerRegion(df):
    #TODO Trouver les régions pour en faire une liste (https://numpy.org/doc/stable/reference/generated/numpy.unique.html)
    liste_regions = np.unique(df['Region'])

    #TODO Pour chacune des régions, trouves les valeurs maximales et le pays associé
    happiness_scores = {}
    for region in liste_regions:
        subset_per_region = CreateSubsetPerRegion(df, region)
        maximum_score = subset_per_region['Happiness score'].max()
        row = subset_per_region.loc[subset_per_region['Happiness score'] == maximum_score]
        country = row['Country name'].iloc[0]
        happiness_scores[country] = maximum_score
        
    #TODO Création d'un dictionnaire en ordre décroissant des valeurs de 'Happiness score'   
    sorted_dict = dict(sorted(happiness_scores.items(), key=lambda item: item[1], reverse=True))

    return sorted_dict

print(ClassMaxScorePerRegion(df))
```

```python
def FindDifferences(df_1, df_2):
    #TODO Reprendre les valeurs obtenues précédemment du pays avec le score maximal par région
    max_score_per_region_1 = ClassMaxScorePerRegion(df_1)
    max_score_per_region_2 = ClassMaxScorePerRegion(df_2)
    #TODO Retourner un dictionnaire des différences en ordre croissant
    print(max_score_per_region_1)
    print(max_score_per_region_2)

    # Vérifier que les structures retournées par la fonction 
    # ClassMaxScorePerRegion(df_1) sont de type dict.
    if (not isinstance(max_score_per_region_1 , dict )) or (not isinstance(max_score_per_region_2, dict)):
        raise TypeError("Les deux entrées doivent être des dictionnaires.")
    dictionnaire_difference = {}
    # Obtenir toutes les clés uniques de chaque dictionnaire.
    ensemble_cles_dic = set(max_score_per_region_1.keys()) | set(max_score_per_region_2.keys())

    # Calculer les différences
    for cle in ensemble_cles_dic:
        valeur1 = max_score_per_region_1.get(cle, 0)
        valeur2 = max_score_per_region_2.get(cle, 0)

        # Vérifier que les entrées du dictionnaire sont numériques
        if (not isinstance(valeur1, (int, float))) or (not isinstance(valeur2, (int, float))):
            raise ValueError(f"Les valeurs qui correspondent à la clé {cle} doivent être des entrées numériques.")
        dictionnaire_difference[cle] = abs(valeur1) - abs(valeur2)

    # Trier le dictionnaire de différence par valeur
    sorted_dict = dict(sorted(dictionnaire_difference.items(), key=lambda item: item[1]))

    return sorted_dict


def PlotComparison(sorted_dict):
    #TODO Faire un bar plot de ces différences
    # Définir le titre et le nom des axes.
    titre = "Différence entre 2016 et 2020 des scores du bonheur en fonction du pays."
    abscisse = "Pays"
    ordonnee = "Différence de scores de bonheur"

    # Extraire les clés et les valeurs dans des listes.
    etiquettes = list(sorted_dict.keys())
    valeurs = list(sorted_dict.values())

    # Créer la figure
    plt.figure(figsize=(12, 7.5))
    plt.bar(etiquettes, valeurs, color='skyblue', edgecolor='blue')

    # Mettre le titre et le nom des axes dans les variables correspondantes de la figure.
    plt.title(titre)
    plt.xlabel(abscisse)
    plt.ylabel(ordonnee)

    # Effectuer une rotation de l'affichage des valeurs de l'abscisse à 90 degrés.
    plt.xticks(rotation=90)
    
    # Ajuster la mise en forme pour que les label s'intègrent bien dans la figure.
    plt.tight_layout()
    
    # Afficher la figure.
    plt.show()
    
sorted_dict = FindDifferences(df, df_2016)
print(sorted_dict)
PlotComparison(sorted_dict)

# Les valeurs qui suivent ont été volontairement affichée, afin de permettre la comparaison.
# En effet, le test de la question 2, n'a pas réussi. Après vérification, j'ai remarqué que
# les clés présentes dans le df_1, mais pas dans le df_2, sont affichés, mais que l'inverse
# n'est pas appliqué. J'ai aussi vérifier les valeurs retournées dans mon cas. La modification
# du fichier de test n'est pas de ma compétence. La présente remarque est adressée en tout respect.
```

## Question 3
Seaborn est une librairie qui permet de plus facilement tracer des graphiques complexes. Souvent, des [subplot](https://matplotlib.org/stable/gallery/subplots_axes_and_figures/subplots_demo.html) via matplotlib sont utilisés afin d'afficher plusieurs graphiques dans la même figure à des fins de comparaisons. Nous nous intéresserons maintenant à deux paramètres particuliers:
1. Healthy life expectancy
2. Happiness score

Vous devez comparer pour les deux années 2016 et 2020 ces deux paramètres. Pour le premier paramètre, vous utiliserez [sns.displot](https://seaborn.pydata.org/generated/seaborn.distplot.html) et pour le second paramètre, vous ferez usage des fonctionnalités de pandas pour le traçage avec [boxplot](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.boxplot.html). Voici un exemple d'affichage:


```python
# Fonction ajoutée afin d'alléger la fonction principale.

# Fonction qui permet de valider les entrées
def validate_dataframe_and_column(df, nom_colonne):
    if not isinstance(df, pd.DataFrame):
        raise TypeError("La structure de donnée doit être une DataFrame de la librairie 'Pandas'.")
    if nom_colonne not in df.columns:
        raise ValueError(f"La colonne '{nom_colonne}' n'existe pas dans la DataFrame.")

    # Eliminer les valeurs NaN de la column
    donnees = df[nom_colonne].dropna()
    if donnees.empty:
        raise ValueError("La colonne ne contient aucune donnée numérique valide.")
    if not pd.api.types.is_numeric_dtype(donnees):
        raise TypeError("La colonne sélectionnée doit contenir des données numériques.")

    return donnees
        
    
def PlotDistributions(df_1, df_2):    
    #TODO Utiliser sns.displot et df.boxplot pour visualiser les deux paramètres demandés
    colonne_histogramme = "Healthy life expectancy"
    colonne_boxplot = "Happiness score"
    annees = [2016, 2020]
    liste_df = [df_1, df_2]

    validate_dataframe_and_column(df_1, colonne_histogramme)
    validate_dataframe_and_column(df_2, colonne_histogramme)

    validate_dataframe_and_column(df_1, colonne_boxplot)
    validate_dataframe_and_column(df_2, colonne_boxplot)

    # Créer la figure subdivisée en quatre ('subplots').
    n_rangees = 2
    n_colonnes = 2
    fig, axs = plt.subplots(n_rangees, n_colonnes, figsize=(12, 8))

    # Histogrammes
    for i in range(n_rangees):
        sns.histplot(x=liste_df[i][colonne_histogramme], bins='auto', kde=False, stat='density', ax=axs[0,i])
        # Ajouter le titre de la figure et des axes
        axs[0,i].set_title(f"{annees[i]}")
        axs[0,i].set_xlabel(colonne_histogramme)
        axs[0,i].set_ylabel("Count")
        axs[0,i].grid("True")

        # Ajouter une courbe tendence sur l'histogramme.
        kde_ax = liste_df[i][colonne_histogramme].plot.kde(ax=axs[0,i], color='blue', linewidth=2)

    # Boxplots
    for j in range(n_rangees):
        liste_df[j].boxplot(column=colonne_boxplot, ax=axs[1,j])
        axs[1,j].set_ylabel("Valeurs")
        axs[1,j].grid(True)

    plt.tight_layout()
    plt.show()


PlotDistributions(df_2016, df)
```

## Question 4
En se concentrant maintenant à nouveau sur les données de **2020**, nous allons essayer de trouver le paramètre le plus corrélé avec le score de bonheur 'Happiness score'. Nous allons prendre des paramètres spécifiques à analyser:
1. Happiness score
2. Logged GDP per capita
3. Social support
4. Healthy life expectancy
5. Freedom to make life choices
6. Generosity
7. Perceptions of corruption

Utilisez un [sns.heatmap](https://seaborn.pydata.org/generated/seaborn.heatmap.html) afin de visualiser ces [corrélations](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.corr.html). Votre heatmap de corrélation devrait ressembler à ce qui est présenté ci-dessous. Attention, ce qui est affiché n'est pas exactement ce que vous allez obtenir puisque ce ne sont pas les même données utilisées. L'exemple n'est qu'indicatif pour la ressemblance visuelle à obtenir.


Pour ce faire, vous allez écrire la fonction FindCorrelations(). En plus d'afficher le heatmap, cette fonction retourne la valeur maximale de corrélation avec le score de bonheur en considérant un paramètre différent de lui-même. En effet, comme vous pouvez le voir à la figure ci-haut le paramètre 'Happiness score' est toujours corrélé à 100% avec lui-même. Nous nous intéresserons donc plutôt à la deuxième valeur de corrélation la plus élevée et le paramètre qui lui est associé. L'affichage final ressemble à ceci et le coefficient de corrélation doit être arrondi au centième près:


```python
LISTE_COLONNES = ['Happiness score', 'Logged GDP per capita', 'Social support', 
    'Healthy life expectancy', 'Freedom to make life choices', 'Generosity', 'Perceptions of corruption']

COLONNE_CIBLE = 'Happiness score'

# Fonction ajoutée pour valider les données dans les colonnes.
def validate_dataframe_and_columns(df, noms_colonnes: list):
    for nom_colonne in noms_colonnes:
        validate_dataframe_and_column(df, nom_colonne)

# Réajuste le fond d'affichage à blanc
matplotlib.rcParams['figure.facecolor'] = 'w'
def FindCorrelations(df):
    #TODO Sélectionner les 7 paramètres à étudier
    # Valider les dataframes
    validate_dataframe_and_columns(df, LISTE_COLONNES)
    # Selectionner les colonnes dans la 'dataframe'
    colonnes_selectionnees = df[LISTE_COLONNES]

    #TODO Faire et afficher un heatmap des facteurs de corrélation
    # Construire la matrice de correlation
    matrice_correlation = colonnes_selectionnees.corr()

    # Selectionner la colonne cible (nom) et les valeurs correspondantes de correlation.
    colonne_cible = COLONNE_CIBLE
    valeurs_correlation_cible = matrice_correlation.loc[colonne_cible] 

    # Obtenir la valeur maximale de correlation pour le paramètre cible, 
    # ainsi que le nom de la colonne correspondante.
    max_corr_val = valeurs_correlation_cible.nlargest(2).iloc[1] 
    max_corr_colonne = valeurs_correlation_cible.nlargest(2).index[1] 

    # Constuire la carte de chaleur
    plt.figure(figsize=(12, 7.5))
    sns.heatmap(
        matrice_correlation,
        cmap='coolwarm',
        annot=True,
        linewidths=0.1
    )

    # Retourner les valeurs demandées.
    return max_corr_val, max_corr_colonne

# Récupérer les valeurs de retours et afficher la carte de chaleur via l'appel de la fonction.
max_corr_val, max_corr_colonne = FindCorrelations(df)
#TODO Afficher la phrase indiquant le coefficient de corrélation le plus élevé avec le paramètre 'Happiness score'
affichage = f"Le paramètre avec un coefficient de corrélation le plus élévé avec Happiness score est {max_corr_colonne} avec {round(max_corr_val, 2)}"
print(affichage)
```
